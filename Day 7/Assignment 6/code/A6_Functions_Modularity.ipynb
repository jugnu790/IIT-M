## Q21. calculate_average()

```python
def calculate_average(numbers):
    if not numbers:
        raise ValueError("The list must not be empty.")
    return sum(numbers) / len(numbers)

numbers = [10, 20, 30, 40]
print(calculate_average(numbers))
```

## Q22. find_maximum()

```python
def find_maximum(numbers):
    if not numbers:
        raise ValueError("The list must not be empty.")

    maximum = numbers[0]
    for number in numbers[1:]:
        if number > maximum:
            maximum = number

    return maximum

numbers = [12, 7, 25, 9, 18]
print(find_maximum(numbers))
```

## Q23. Modular data program

```python
def read_data():
    return [10, 20, 30, 40, 50]

def calculate_mean(numbers):
    if not numbers:
        raise ValueError("The list must not be empty.")
    return sum(numbers) / len(numbers)

def calculate_maximum(numbers):
    if not numbers:
        raise ValueError("The list must not be empty.")

    maximum = numbers[0]
    for number in numbers[1:]:
        if number > maximum:
            maximum = number

    return maximum

def display_results(numbers):
    print("Data:", numbers)
    print("Mean:", calculate_mean(numbers))
    print("Maximum:", calculate_maximum(numbers))

numbers = read_data()
display_results(numbers)
```

## Q24. count_even()

```python
def count_even(numbers):
    count = 0

    for number in numbers:
        if number % 2 == 0:
            count += 1

    return count

numbers = [1, 2, 3, 4, 6, 9]
print(count_even(numbers))
```

## Q25. Classification accuracy

```python
def calculate_accuracy(y_true, y_pred):
    if len(y_true) != len(y_pred):
        raise ValueError("y_true and y_pred must have the same length.")
    if not y_true:
        raise ValueError("Input lists must not be empty.")

    correct = 0
    for actual, predicted in zip(y_true, y_pred):
        if actual == predicted:
            correct += 1

    return correct / len(y_true)

def display_result(accuracy):
    print(f"Accuracy: {accuracy:.2%}")

y_true = [1, 0, 1, 1, 0]
y_pred = [1, 1, 1, 0, 0]

accuracy = calculate_accuracy(y_true, y_pred)
display_result(accuracy)
```

## Q26. Mean and standard deviation

```python
def mean_and_std(numbers):
    if not numbers:
        raise ValueError("The list must not be empty.")

    mean = sum(numbers) / len(numbers)
    variance = sum((x - mean) ** 2 for x in numbers) / len(numbers)
    std = variance ** 0.5

    return mean, std

numbers = [2, 4, 4, 4, 5, 5, 7, 9]
mean, std = mean_and_std(numbers)

print("Mean:", mean)
print("Standard deviation:", std)
```

## Q27. Min-max normalization

```python
def normalize(numbers):
    if not numbers:
        raise ValueError("The list must not be empty.")

    min_value = min(numbers)
    max_value = max(numbers)

    if max_value == min_value:
        return [0.0] * len(numbers)

    return [
        (x - min_value) / (max_value - min_value)
        for x in numbers
    ]

numbers = [10, 20, 30, 40, 50]
print(normalize(numbers))
```

## Q28. Modular preprocessing pipeline

```python
def load_data():
    return [10, None, 20, 30, None, 40]

def remove_missing_values(data):
    return [x for x in data if x is not None]

def normalize_data(data):
    if not data:
        return []

    min_value = min(data)
    max_value = max(data)

    if max_value == min_value:
        return [0.0] * len(data)

    return [
        (x - min_value) / (max_value - min_value)
        for x in data
    ]

def display_data(data):
    print("Processed data:", data)

data = load_data()
data = remove_missing_values(data)
data = normalize_data(data)
display_data(data)
```

## Q29. Euclidean distance

```python
def euclidean_distance(point1, point2):
    if len(point1) != len(point2):
        raise ValueError("Points must have the same number of dimensions.")

    squared_distance = 0
    for a, b in zip(point1, point2):
        squared_distance += (a - b) ** 2

    return squared_distance ** 0.5

point1 = [1, 2, 3]
point2 = [4, 6, 3]

print(euclidean_distance(point1, point2))
```
