## Task 4 — Indexing a Tensor

## Task 5 — Slicing

```python
X = np.arange(60).reshape(3, 4, 5)

first_block = X[0]
first_row_every_block = X[:, 0, :]
third_element_every_row_every_block = X[:, :, 2]
last_two_elements = X[:, :, -2:]

print("X shape:", X.shape)
print("\nFirst block X[0]:\n", first_block)
print("\nFirst row from every block X[:, 0, :]:\n", first_row_every_block)
print("\nThird element from every row and block X[:, :, 2]:\n", third_element_every_row_every_block)
print("\nLast two elements X[:, :, -2:]:\n", last_two_elements)

print("\nResulting shapes:")
print("First block:", first_block.shape)
print("First row from every block:", first_row_every_block.shape)
print("Third element from every row/block:", third_element_every_row_every_block.shape)
print("Last two elements:", last_two_elements.shape)
```

## Task 8 — Reshape

```python
X = np.arange(24)

reshapes = [(4, 6), (2, 3, 4), (6, 4), (24,)]

for shape in reshapes:
    Y = X.reshape(shape)
    print("=" * 50)
    print("Shape:", shape)
    print("Tensor:\n", Y)
    print("Number of dimensions:", Y.ndim)
    print("Number of elements:", Y.size)

print("\nExplanation:")
print("Reshaping changes how the same elements are organized into axes.")
print("It cannot change the total number of elements; therefore every valid reshape must contain 24 elements.")
```

## Task 9 — Reshape vs Transpose

```python
X = np.arange(24).reshape(2, 3, 4)

reshape_result = X.reshape(3, 2, 4)
transpose_result = X.transpose(1, 0, 2)

print("Original X:\n", X)
print("\nX.reshape(3, 2, 4):\n", reshape_result)
print("\nX.transpose(1, 0, 2):\n", transpose_result)

print("\nSame shape?", reshape_result.shape == transpose_result.shape)
print("Same values in the same positions?", np.array_equal(reshape_result, transpose_result))

print("\nConclusion:")
print("No. Reshape changes the grouping/order interpretation of the existing sequence.")
print("Transpose permutes existing axes while preserving their axis-wise relationships.")
```

## Task 10 — Broadcasting

```python
X = np.ones((3, 4))
b = np.array([10, 20, 30, 40])

addition = X + b
multiplication = X * b

print("X + b:\n", addition)
print("\nX * b:\n", multiplication)
print("\nShapes:")
print("X:", X.shape)
print("b:", b.shape)
print("X + b:", addition.shape)
print("X * b:", multiplication.shape)

print("\nWhy it works:")
print("The (4,) vector matches the final dimension of (3, 4).")
print("NumPy broadcasts b across all 3 rows.")
```

## Task 12 — Tensor Aggregation

```python
X = rng.normal(size=(100, 20, 5))

mean_across_time = X.mean(axis=1)
mean_across_sensors = X.mean(axis=2)
mean_across_machines = X.mean(axis=0)
max_sensor_value_across_time = X.max(axis=1)
std_across_time = X.std(axis=1)

print("X shape:", X.shape)
print("\nPredicted and actual result shapes:")
print("1. Mean across time:        ", mean_across_time.shape, " (100, 5)")
print("2. Mean across sensors:     ", mean_across_sensors.shape, " (100, 20)")
print("3. Mean across machines:    ", mean_across_machines.shape, " (20, 5)")
print("4. Max across time:         ", max_sensor_value_across_time.shape, " (100, 5)")
print("5. Std across time:         ", std_across_time.shape, " (100, 5)")
```

## Task 13 — Normalize a Tensor

```python
X = rng.normal(size=(100, 20, 5))

# Mean and standard deviation for each sensor across machines and time.
sensor_mean = X.mean(axis=(0, 1), keepdims=True)
sensor_std = X.std(axis=(0, 1), keepdims=True)

X_normalized = (X - sensor_mean) / sensor_std

verified_mean = X_normalized.mean(axis=(0, 1))
verified_std = X_normalized.std(axis=(0, 1))

print("Original shape:", X.shape)
print("Sensor mean shape:", sensor_mean.shape)
print("Sensor std shape:", sensor_std.shape)
print("\nNormalized mean for each sensor:", verified_mean)
print("Normalized std for each sensor:", verified_std)

print("\nMean approximately 0?", np.allclose(verified_mean, 0, atol=1e-7))
print("Std approximately 1?", np.allclose(verified_std, 1, atol=1e-7))
```

## Task 15 — Concatenation

```python
visual_features = rng.normal(size=(32, 128))
sensor_features = rng.normal(size=(32, 64))

combined = np.concatenate([visual_features, sensor_features], axis=1)

print("Visual features:", visual_features.shape)
print("Sensor features:", sensor_features.shape)
print("Combined features:", combined.shape)

print("\nWhy concatenate along feature dimension?")
print("Both tensors have the same batch dimension (32).")
print("Their feature dimensions (128 and 64) can be joined along axis 1.")
print("Dimensions being concatenated must be compatible on the other axes.")
```

## Task 16 — Practical Mini Challenge: Sensor Dataset

```python
X = rng.normal(size=(500, 60, 8))

print("Dataset shape:", X.shape)

# 1. Machine 10 (zero-based index 9)
machine_10 = X[9]

# 2. First 20 time steps from all machines
first_20_time_steps = X[:, :20, :]

# 3. Sensor 5 from all machines (zero-based index 4)
sensor_5 = X[:, :, 4]

# 4. Average value of every sensor across machines and time
average_every_sensor = X.mean(axis=(0, 1))

# 5. Average value at every time step across machines and sensors
average_every_time_step = X.mean(axis=(0, 2))

# 6. Machine with highest average sensor reading
machine_means = X.mean(axis=(1, 2))
highest_machine_index = np.argmax(machine_means)
highest_machine_number = highest_machine_index + 1

# 7. Normalize each sensor across machines and time
sensor_mean = X.mean(axis=(0, 1), keepdims=True)
sensor_std = X.std(axis=(0, 1), keepdims=True)
X_normalized = (X - sensor_mean) / sensor_std

# 8. Train/test split
X_train = X[:400]
X_test = X[400:]

print("\n1. Machine 10 shape:", machine_10.shape)
print("2. First 20 time steps shape:", first_20_time_steps.shape)
print("3. Sensor 5 shape:", sensor_5.shape)
print("4. Average of every sensor:", average_every_sensor)
print("5. Average at every time step shape:", average_every_time_step.shape)
print("6. Machine with highest average sensor reading:", highest_machine_number)
print("   Highest average:", machine_means[highest_machine_index])
print("7. Normalized tensor shape:", X_normalized.shape)
print("8. Training shape:", X_train.shape)
print("   Testing shape:", X_test.shape)
```
